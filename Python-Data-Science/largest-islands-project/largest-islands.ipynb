```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('largest-islands.csv')
df.head()
```

### Project Ideas

- What are the 10 largest islands in the tropics?

- What are the largest islands in each `region`?

- Create a line graph with `area` on the y-axis and `rank` on the x-axis. The data should be ordered by `rank`, from largest to smallest.

- What islands are composed of multiple countries?
	- Hint: try `countries.str.contains(",")`

```python
# YOUR CODE HERE (add more cells as needed)
df.columns
df.head()
```

```python
tropical_islands = df.query("climate == 'tropics'")

tropical_islands.sort_values("rank").head(10)
```

```python
largest_by_region = df.sort_values("rank").groupby("region").first()

largest_by_region
```

```python
islands_sorted = df.sort_values("rank")

plt.plot(islands_sorted["rank"], islands_sorted["area"])

plt.xlabel("Rank")
plt.ylabel("Area")
plt.show()
```
